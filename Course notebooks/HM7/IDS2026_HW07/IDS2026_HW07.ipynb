**e. Identify potential constraints.**

These could be time limitations, data privacy concerns, regulatory requirements, technical limitations, and ethical considerations.

**Answer:** Data Constraints:

Several data constraints must be considered during analysis. First, focusing exclusively on qualifying data from a single race (such as the Singapore GP) severely limits the number of official representative laps, which increases the risk of session-specific variance. However, restricting the dataset to a single session ensures consistent environmental baseline conditions, including track temperature, weather, wind speed, and track evolution. Introducing data from multiple sessions or tracks would add numerous unknown variables that could distort the results.

Second, relying solely on publicly available FastF1 telemetry means we lack internal team data, creating unobserved confounders. Crucial factors such as exact fuel loads, engine modes, tire age, setup changes, or physical car damage are not captured in public streams. Without this internal context, there is a risk of variable bias—for example, mistaking a damaged car or a specific engine cooling issue for a poor driving input or an ineffective aerodynamic upgrade, ultimately leading to misleading conclusions.

**f. EXTRA. Possible data errors**

With real world data there is so much data, that is possible to draw wrong conclusions, I want to name some errors that I am going to try to avoid in my analysis, but I might not have the data always for.

**Answer:** Data Errors:

- Error 1: Driver's health. Because this sport is ruthless, missing a session for health reasons is extremely rare—every session a driver skips is a missed opportunity to improve or demonstrate their ability. As a result, drivers may sometimes compete while feeling under the weather, which directly affects their driving performance and telemetry outputs.

- Error 2: High temperature. In Singapore, ambient temperatures normally hover around 30°C, and drivers sit directly in front of power units that operate at extreme temperatures. The physical strain of operating a Formula 1 car at over 300 km/h in these conditions is immense. While driver cooling vests provide some relief, heat remains a significant performance factor for both the driver and the thermal management of the car.

- Error 3: Minor car damage. Due to strict budget caps and component regulations, light structural or aerodynamic damage does not automatically trigger a part replacement. Teams frequently use high-speed tape for minor repairs to avoid consuming valuable financial and component quotas. Consequently, cars may run with slight structural damage or wear originating from earlier practice sessions or previous race weekends.

- Error 4: Undocumented specifications. While F1 teams must report new component introductions, they are not required to re-declare components used in previous races, nor are they obligated to publicly state if they have reverted to an older specification setup. Furthermore, track-specific adaptations—such as high-downforce rear wing configurations for Singapore—make it difficult to verify the exact baseline spec from raw telemetry alone.

- Error 5: Missing tyre usage data. Each team receives a fixed allocation of tyre compounds (soft, medium, hard). However, public telemetry often lacks precise context regarding tyre degradation, previous stint history, exact age (in laps), and thermal cycles—such as whether a tyre was subjected to an aggressive out-lap or aborted after a single push lap.

- Error 6: Traffic and track incidents. On tight street circuits like Singapore, drivers frequently have to abandon hot laps or slow down due to traffic or yellow flags. If an aborted lap or heavily compromised sector is accidentally included in the regression model, it corrupts the true performance baseline.

- Error 7: Engine status, mapping modes, and component limits. Public FastF1 telemetry does not report internal engine sensor data, such as power unit mapping modes, hybrid deployment settings, oil pressure, or internal temperatures. Additionally, telemetry does not track component usage counts (e.g., whether a driver is using their 1st or 4th internal combustion engine of the season). As a result, if a team de-tunes an engine to preserve reliability, experiences hidden thermal clipping, or runs an old high-mileage power unit during practice/qualifying, the resulting drop in straight-line speed could be falsely attributed to poor aerodynamic efficiency or driver error.
